```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from tensorflow import keras
```

```python
(X_train, y_train) , (X_test, y_test) = keras.datasets.mnist.load_data()
(X_train, y_train) , (X_test, y_test) = keras.datasets.mnist.load_data()
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
# A standard Dense (fully connected) layer consists of individual neurons. Each neuron in the first hidden layer must connect to every single input
# feature.Standard dense layers do not understand spatial grids (rows and columns); they only understand a flat list of features. By flattening a
# 28 \times 28 image, you transform it into a 1D array of 784 individual pixel values (28 \times 28 = 784). This maps perfectly to an input layer
# with 784 input nodes.
X_train_flattened = X_train.reshape(len(X_train), 784)
X_test_flattened = X_test.reshape(len(X_test), 784)
```

```python
X_train_flattened.shape
```

```python
X_train_flattened[0]
```

```python
from tensorflow.keras.layers import Dense,Input
```

```python
model = keras.Sequential([
    Input(shape=(784,)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(X_train_flattened, y_train, epochs=5)
```

```python
model.evaluate(X_test_flattened, y_test)
```

```python
model = keras.Sequential([
    keras.layers.Dense(100, input_shape=(784,), activation='relu'),
    keras.layers.Dense(10, activation='sigmoid')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(X_train_flattened, y_train, epochs=5)
```

```python
model.evaluate(X_test_flattened,y_test)
```
